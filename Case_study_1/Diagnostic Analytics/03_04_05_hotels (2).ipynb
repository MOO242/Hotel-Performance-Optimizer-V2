```python
yearly_room_revenue = (
    read_table("fact_reservations")[
        read_table("fact_reservations")["booking_status"] == "Checked Out"
    ]
    .merge(
        read_table("dim_date"),
        how="inner",
        left_on="booking_date_key",
        right_on="date_key",
    )
    .merge(
        read_table("dim_hotels"),
        how="left",
        left_on="property_id",
        right_on="property_id",
    )
    .groupby(
        [
            "year",
            "property_id",
            "property_name",
        ]
    )["room_revenue"]
    .sum()
).reset_index()


yearly_room_revenue
```

```python
yearly_room_available = (
    read_table("fact_room_inventory")
    .merge(
        read_table("dim_date"),
        how="inner",
        left_on="date_key",
        right_on="date_key",
    )
    .merge(
        read_table("dim_hotels"),
        how="left",
        left_on="property_id",
        right_on="property_id",
    )
    .groupby(
        [
            "year",
            "property_id",
            "property_name",
        ],
        as_index=False,
    )
).agg(room_available=("rooms_available", "sum"), room_sold=("rooms_sold", "sum"))


yearly_room_available
```

```python
# Aggregate availability at the same merge grain to avoid a many-to-many merge.


yearly_revpar = yearly_room_revenue.merge(
    yearly_room_available,
    how="inner",
    on=["year", "property_id", "property_name"],
    validate="one_to_one",
)


yearly_revpar["revpar"] = (
    yearly_revpar["room_revenue"] / yearly_revpar["room_available"].replace(0, np.nan)
).round(2)


yearly_revpar
```

```python
# validation

before = len(yearly_room_revenue)
# ... merge ...
assert (
    len(yearly_revpar) == before
), f"Row count changed: {before} → {len(yearly_revpar)}"
assert yearly_revpar["room_available"].isna().sum() == 0, "Unmatched properties"
```

```python
yearly_revpar["adr"] = (
    yearly_revpar["room_revenue"] / yearly_revpar["room_sold"].replace(0, np.nan)
).round(2)

yearly_revpar["occupancy"] = (
    yearly_revpar["room_sold"]
    / yearly_revpar["room_available"].replace(0, np.nan)
    * 100
).round(2)
```

### 2 - Which properties contributed most to revenue changes?

```python
top10 = (
    yearly_revpar.sort_values(["year", "room_revenue"], ascending=[True, False])
    .groupby("year")
    .head(10)
)
```

```python
top10
```

### 3 - Which hotels contributed most to Revenue Growth %?

```python
top10 = top10.sort_values(["year", "property_name"])

top10["YOY_GROWTH_PCT"] = (
    top10.groupby("property_name")["revpar"].pct_change() * 100
).round(2)

top10
```

### 4- Which hotels experienced the largest revenue declines?

```python
top10_declines = (
    yearly_revpar.sort_values(["year", "room_revenue"], ascending=[True, True])
    .groupby("year")
    .head(10)
)


top10_declines = top10_declines.sort_values(["year", "property_name"])

top10_declines["YOY_GROWTH_PCT"] = (
    top10_declines.groupby("property_name")["revpar"].pct_change() * 100
).round(2)

top10_declines
```

```python
# Step 1: compute YoY revenue change on the FULL dataset first
yearly_revpar = yearly_revpar.sort_values(["property_id", "year"])

yearly_revpar["revenue_yoy_pct"] = (
    yearly_revpar.groupby("property_id")["room_revenue"].pct_change() * 100
).round(2)

# Step 2: drop rows with no prior year to compare (first year per hotel)
declines = yearly_revpar.dropna(subset=["revenue_yoy_pct"])

# Step 3: NOW rank by the change itself — most negative = biggest decline
top10_declines = declines.sort_values("revenue_yoy_pct", ascending=True).head(10)

# Or per-year, if you want "worst 10 decliners within each year":
top10_declines_by_year = (
    declines.sort_values(["year", "revenue_yoy_pct"], ascending=[True, True])
    .groupby("year")
    .head(10)
)

top10_declines_by_year
```
