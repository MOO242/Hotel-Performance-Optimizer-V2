# Snowflake connection

```python
import os
from dotenv import load_dotenv
import snowflake.connector
load_dotenv()

conn = snowflake.connector.connect(
    
account = os.getenv("SNOWFLAKE_ACCOUNT"),
user = os.getenv("SNOWFLAKE_USER"),
password = os.getenv("SNOWFLAKE_PASSWORD"),
warehouse = os.getenv("SNOWFLAKE_WAREHOUSE"),
database = os.getenv("SNOWFLAKE_DATABASE"),
schema = os.getenv("SNOWFLAKE_SCHEMA")    

)

def run_query(sql):
    cur = conn.cursor()
    cur.execute (sql)
    return  cur.fetch_pandas_all()
```

# Sanity-Check 

```python
df_bookings  = run_query("SELECT * FROM FACT_BOOKINGS")
print(df_bookings.shape)

df_aggregated   = run_query("SELECT * FROM FACT_AGGREGATED_BOOKINGS")
print(df_aggregated.shape)

df_hotels   = run_query("SELECT * FROM DIM_HOTELS")
print(df_hotels.shape)

df_rooms  = run_query("SELECT * FROM DIM_ROOMS")
print(df_rooms.shape)

df_customers  = run_query("SELECT * FROM DIM_CUSTOMERS")
print(df_customers.shape)

df_date  = run_query("SELECT * FROM DIM_DATE")
print(df_date.shape)
```
